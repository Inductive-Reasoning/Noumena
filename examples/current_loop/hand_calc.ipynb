```python
from scipy.special import ellipk, ellipe
import math
```

```python
def A_phi(r, z):
    mu_0 = 4*math.pi*1e-7
    a = 0.1
    I = 1.0

    k = math.sqrt(4*a*r/((a+r)**2+z**2))
    return mu_0 * I / (math.pi * k) * math.sqrt(a/r) * ((1-k**2/2)*ellipk(k)-ellipe(k))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the grid
r = np.linspace(0.05, 0.5, 100) # Start from 0.001 to avoid division by zero at r=0
z = np.linspace(-0.5, 0.5, 100)
R, Z = np.meshgrid(r, z)

# Vectorize the A_phi function to work with numpy arrays
A_phi_vec = np.vectorize(A_phi)

# Calculate A_phi over the grid
A = A_phi_vec(R, Z)

# Plot the results
plt.figure(figsize=(10, 8))
cp = plt.contourf(R, Z, A, levels=50, cmap='viridis')
plt.colorbar(cp, label='A_phi')
plt.title('Magnetic Vector Potential A_phi')
plt.xlabel('r')
plt.ylabel('z')
plt.axis('equal')
plt.show()
```

## Self-inductance of the loop

The `A_phi` expression above is for a *filamentary* loop, so it diverges as
`r -> a` and cannot give a self-inductance on its own. For a conductor of
finite cross-section the singularity is removed by replacing the filament
radius with the geometric mean distance (GMD) of the section. For a square of
side `w` the GMD is `r_eq = 0.2235 * (w + h) = 0.447 * w`, and because the GMD
already accounts for the internal inductance, Grover's thin-ring formula is
used in its `-2` form:

    L = mu_0 * a * (ln(8a / r_eq) - 2)

This is the reference value asserted by the solver regression tests
"Magnetostatic loop inductance matches the analytic ring value" and its
magnetoquasistatic counterpart in `test/test_solvers.cpp`.


```python
mu_0 = 4 * math.pi * 1e-7
a = 0.1        # loop radius [m]
w = 0.002      # square cross-section side [m]

r_eq = 0.2235 * (w + w)          # GMD of a square cross-section
L = mu_0 * a * (math.log(8 * a / r_eq) - 2)

print("r_eq = {:.6e} m".format(r_eq))
print("L    = {:.6e} H  ({:.2f} nH)".format(L, L * 1e9))

# Values reported by noumena on the same geometry with
# analysis_type = "coupling_matrix" and the far boundary at D/a = 40.
results = [
    ("magnetostatic", 6.023565e-07),
    ("MQS @ 0.1 Hz", 6.023565e-07),
]
for name, value in results:
    print("{:>16}: {:.6e} H   error = {:+.2f} %".format(
        name, value, 100.0 * (value - L) / L))

# The residual is far-field truncation, not the GMD approximation: imposing
# A_phi = 0 at a finite radius biases L low, with the error decaying as 1/D.
# Extrapolating the domain sweep to D -> infinity gives 6.0265e-07 H, within
# -0.013% of the formula above. See docs/open_boundary.md.
for domain_over_a, err in [(5, -0.49), (10, -0.20), (40, -0.06), (160, -0.03)]:
    print("D/a = {:>4}: {:+.2f} %".format(domain_over_a, err))
```
